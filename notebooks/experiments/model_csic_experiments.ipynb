# Experimentos — Model CSIC (CSIC 2010)

**Objetivo:** Experimentar con 4 modelos de ML para detección de ataques web.
Usa la función `train_model_a()` del módulo `src.mlsec.data.train_model_a`.

**Métricas objetivo:**
- Recall >= 0.95
- Precision >= 0.85

**MLflow:**
- Model Registry: `model-csic`
- Experiment: `model-csic`
- Tracking: `http://localhost:5081`

```python
import os
import sys
from pathlib import Path

# Agregar src al path
ROOT = Path.cwd()
while not (ROOT / 'mkdocs.yml').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import mlflow
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Config — v5 con ratio features
FEATURES_PATH = ROOT / 'data' / 'processed' / 'csic2010' / 'features_v5.parquet'
MLFLOW_TRACKING_URI = os.environ.get('MLFLOW_TRACKING_URI', 'http://localhost:5081')
MODEL_NAME = 'model-csic'  # Nombre del modelo en MLflow Registry

print(f'ROOT: {ROOT}')
print(f'Features: {FEATURES_PATH}')
print(f'MLflow: {MLFLOW_TRACKING_URI}')
print(f'Model name: {MODEL_NAME}')
print(f'Features existe: {FEATURES_PATH.exists()}')
```

## 1. Ejecutar experimentos

Se entrenan 4 modelos (LogisticRegression, RandomForest, XGBoost, LightGBM)
usando la función `train_model_a()` del módulo compartido.

Todos los resultados se loggean automáticamente a MLflow.

```python
from src.mlsec.data.train_model_a import train_model_a

# Ejecutar experimentos — esto entrena los 4 modelos y los loggea a MLflow
results = train_model_a(
    features_path=FEATURES_PATH,
    model_name=MODEL_NAME,
    mlflow_tracking_uri=MLFLOW_TRACKING_URI,
    skip_promote=True,  # No promover automáticamente en experimentación
)

print('\n=== RESUMEN ===')
print(f"Modelos entrenados: {list(results['results'].keys())}")
for name, m in results['results'].items():
    recall_ok = '✅' if m['test_recall'] >= 0.95 else '❌'
    precision_ok = '✅' if m['test_precision'] >= 0.85 else '❌'
    print(f"{name}: Recall={m['test_recall']:.4f} {recall_ok} | Precision={m['test_precision']:.4f} {precision_ok} | ROC-AUC={m['test_auc']:.4f}")
```

## 2. Comparativa de modelos

```python
# Tabla comparativa
df_results = pd.DataFrame({
    name: {
        'ROC-AUC (test)': m['test_auc'],
        'Recall (test)': m['test_recall'],
        'Precision (test)': m['test_precision'],
        'FP': m['fp'],
        'Threshold': m['threshold'],
        'Gap precision': m['gap'],
        'Recall OK?': m['test_recall'] >= 0.95,
        'Precision OK?': m['test_precision'] >= 0.85,
    }
    for name, m in results['results'].items()
}).T

print(df_results.to_string())
print()
print('Criterios: Recall >= 0.95 | Precision >= 0.85')
```

## 3. Análisis del mejor modelo

Elegimos el mejor modelo según:
1. Recall >= 0.95 (cumplir criterio mínimo)
2. Mayor Precision (menos falsas alarmas)

```python
# Encontrar mejor modelo: Recall >= 0.95, luego mayor Precision
best = None
best_score = 0
for name, m in results['results'].items():
    if m['test_recall'] >= 0.95 and m['test_precision'] > best_score:
        best_score = m['test_precision']
        best = (name, m)

if best:
    name, m = best
    print(f'Mejor modelo: {name}')
    print(f'  Recall:    {m["test_recall"]:.4f} ✅')
    print(f'  Precision: {m["test_precision"]:.4f}')
    print(f'  ROC-AUC:   {m["test_auc"]:.4f}')
    print(f'  FP:        {m["fp"]}')
    print(f'  Threshold: {m["threshold"]:.4f}')
else:
    print('Ningún modelo cumple Recall >= 0.95')
```

## 4. Conclusiones y próximos pasos

### Hallazgos

```python
# Analizar gap precision
print('=== Análisis de Gap Precision ===')
for name, m in results['results'].items():
    gap_ok = '✅' if m['gap'] <= 0.05 else '❌'
    print(f"{name}: Gap={m['gap']:.4f} {gap_ok}")

print()
print('=== Próximos pasos ===')
recall_candidates = [name for name, m in results['results'].items() if m['test_recall'] >= 0.95]
if recall_candidates:
    print(f'Modelos con Recall >= 0.95: {recall_candidates}')
    best_recall = max(recall_candidates, key=lambda n: results['results'][n]['test_precision'])
    print(f'Mejor Precision entre ellos: {best_recall} ({results["results"][best_recall]["test_precision"]:.4f})')
    precision_gap = 0.85 - results['results'][best_recall]['test_precision']
    if precision_gap > 0:
        print(f'Gap para alcanzar Precision 0.85: {precision_gap:.4f}')
else:
    print('Ningún modelo cumple Recall >= 0.95 — revisar features o hiperparámetros')
```
